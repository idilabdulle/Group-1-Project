## Standardising/scaling features

Feature scaling is not an absolute necessity for linear regression, with one input feature (X), however we are including it here as it can be show to improve the model and is a demonstration of a good workflow.

```python
# Use standard scaler - already imported
scaler = StandardScaler()

# Scaling the X train and test data
scaled_X_train = scaler.fit_transform(X_train)
scaled_X_test = scaler.transform(X_test)
```

## K-Fold Cross Validation

K-Fold cross validation on training data. This is more of a necessity for complex models, such as random forest or neural network, however this will provide us with a more accurate understanding of how the model performs. This can also be used as a solid base of understanding, if we then wanted to then compare this model to other, more complex, models later in the process.

```python
# Define your K-Fold Cross Validation 'settings'
kf = KFold(n_splits=5, random_state=42, shuffle=True)

# Create your model
model = LinearRegression()

# Utilise K-Fold Cross Validation to assess the model
# First define the metrics you need
scoring = ['r2', 'neg_mean_absolute_error',
           'neg_root_mean_squared_error']
kf_scores = cross_val_score(
    model, X_train, y_train, cv=kf, scoring=scoring)

print('Cross validation scores:\n')
print(f'r2: {kf_scores['r2'].mean()}')
print(f'MAE: {kf_scores['neg_mean_absolute_error'].mean()}')
print(f'RMSE: {kf_scores['neg_root_mean_squared_error'].mean()}')
```

## Train the model

```python
# Fit your model to the entire training split of data 
#(rather than 80% splits as seen with the K-Fold Cross Validation above)
model.fit(scaled_X_train, y_train)
```

## Test the model
 - Test the model (using our test data split)
 - Including metrics: R2, MAE, RMSE
 - Testing at this stage can, additionally, impact model selection. For instance, comparing to multiple linear regression or other more complex models

```python
# Use the model to make predictions on your test data
y_pred = model.predict(scaled_X_test)

# Assess the model using the chosen metrics
# R2, MAE, RMSE
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = root_mean_squared_error(y_test, y_pred)
print(f'Test metrics:\nR2: {r2: .5f}\nMAE: {mae: .5f}\nRMSE: {rmse: .5f}')
```
